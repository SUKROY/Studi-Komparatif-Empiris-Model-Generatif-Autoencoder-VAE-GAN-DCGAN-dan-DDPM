```python

import sys
import os
import torch

sys.path.append(os.path.abspath(".."))

from models.vanilla_autoencoder import VanillaAutoencoder
from models.sparse_autoencoder import SparseAutoencoder
from models.vae import VariationalAutoencoder
from models.vanilla_gan import Generator, Discriminator
from models.dcgan import DCGANGenerator, DCGANDiscriminator
from models.ddpm import DDPMNetwork

print("All model imports successful!")
```

```python

batch_size = 4
x = torch.rand(batch_size, 1, 28, 28)

# Autoencoders
ae = VanillaAutoencoder()
print("Vanilla AE:", ae(x).shape)

sparse_ae = SparseAutoencoder()
reconstructed, z = sparse_ae(x)
print("Sparse AE:", reconstructed.shape, z.shape)

vae = VariationalAutoencoder()
reconstructed, mu, logvar = vae(x)
print("VAE:", reconstructed.shape, mu.shape, logvar.shape)

# Vanilla GAN
noise = torch.randn(batch_size, 100)
generator = Generator()
discriminator = Discriminator()

fake = generator(noise)
print("Vanilla GAN Generator:", fake.shape)
print("Vanilla GAN Discriminator:", discriminator(fake).shape)

# DCGAN
dcgan_generator = DCGANGenerator()
dcgan_discriminator = DCGANDiscriminator()

fake = dcgan_generator(noise)
print("DCGAN Generator:", fake.shape)
print("DCGAN Discriminator:", dcgan_discriminator(fake).shape)

# DDPM
ddpm = DDPMNetwork()
t = torch.randint(0, 300, (batch_size,))
predicted_noise = ddpm(x, t)
print("DDPM:", predicted_noise.shape)
```
